## Konsep dan model

Segmentasi memisahkan area objek. Metode Otsu memilih ambang berdasarkan variasi antar-kelas. Untuk pengukuran fisik, luas piksel harus dikalibrasi dengan skala ($A_{nyata}=A_{piksel}\,s^2$).

## Binarisasi dan labeling

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data, filters, morphology, measure
img = data.coins()
threshold=filters.threshold_otsu(img)
mask=img>threshold
mask=morphology.remove_small_objects(mask,min_size=80)
labeled=measure.label(mask)
props=measure.regionprops(labeled)
print('Ambang:',threshold,'Jumlah region:',len(props))
```

## Tampilkan objek terukur

```python
from skimage.color import label2rgb
fig,ax=plt.subplots(1,2,figsize=(9,4))
ax[0].imshow(img,cmap='gray'); ax[0].set_title('Citra masukan')
ax[1].imshow(label2rgb(labeled,bg_label=0));ax[1].set_title('Region tersegmentasi')
for p in sorted(props,key=lambda p:p.area,reverse=True)[:10]:
    y,x=p.centroid
    ax[1].text(x,y,str(p.area),color='white',fontsize=7,ha='center')
for a in ax:a.axis('off')
plt.show()
```

## Luas dan bentuk

```python
area=np.array([p.area for p in props]); solidity=np.array([p.solidity for p in props]);
print('Luas (piksel kuadrat) 5 terbesar:',sorted(area,reverse=True)[:5])
print('Soliditas rata-rata:',round(float(solidity.mean()),3))
```
