## Konsep dan model

Tepi ideal merupakan transisi intensitas tajam. Sobel mengestimasi gradien $|\nabla I|=\sqrt{I_x^2+I_y^2}$. Erosi/dilasi serta opening/closing bekerja pada objek biner dan membutuhkan keputusan ukuran elemen struktur.

## Deteksi tepi

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data, filters, feature
img=data.camera().astype(float)/255
sobel=filters.sobel(img)
canny=feature.canny(img, sigma=1.5)
fig,ax=plt.subplots(1,3,figsize=(10,3))
for a, x, t in zip(ax,[img,sobel,canny],['Input','Sobel','Canny']):
    a.imshow(x,cmap='gray');a.set_title(t);a.axis('off')
plt.show()
```

## Morfologi pada koin

```python
from skimage import morphology, measure
coins=data.coins()
mask=coins>filters.threshold_otsu(coins)
opened=morphology.opening(mask,morphology.disk(2))
closed=morphology.closing(opened,morphology.disk(3))
fig,ax=plt.subplots(1,3,figsize=(10,3))
for a,x,t in zip(ax,[mask,opened,closed],['Ambang Otsu','Opening','Closing']):
    a.imshow(x,cmap='gray');a.set_title(t);a.axis('off')
plt.show()
print('Komponen hasil:',measure.label(closed).max())
```
