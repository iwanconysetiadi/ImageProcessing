## Konsep dan model

Transformasi Fourier 2-D $F(u,v)=\sum_r\sum_c I(r,c)e^{-j2\pi(ur/M+vc/N)}$. Magnitudo spektrum merepresentasikan kontribusi frekuensi spasial, bukan nilai piksel langsung. Filter frekuensi tinggi mempertajam perubahan, tetapi dapat memperbesar noise.

## Spektrum citra

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
img=data.camera().astype(float)/255
F=np.fft.fftshift(np.fft.fft2(img))
S=np.log1p(np.abs(F))
fig,ax=plt.subplots(1,2,figsize=(8,4));ax[0].imshow(img,cmap='gray');ax[1].imshow(S,cmap='magma')
ax[0].set_title('Input');ax[1].set_title('Log spektrum')
for a in ax:a.axis('off')
plt.show()
```

## Filter low pass Gaussian di frekuensi

```python
h,w=img.shape
y,x=np.ogrid[-h//2:h-h//2,-w//2:w-w//2]
r2=x*x+y*y
sigma_f=20
H=np.exp(-r2/(2*sigma_f**2))
low=np.fft.ifft2(np.fft.ifftshift(F*H)).real
fig,ax=plt.subplots(1,3,figsize=(10,3))
for a,x,t in zip(ax,[img,H,low],['Asli','Filter |H|','Low-pass']):
    a.imshow(x,cmap='gray');a.set_title(t);a.axis('off')
plt.show()
print('Energi perubahan (MSE):',np.mean((img-low)**2))
```
