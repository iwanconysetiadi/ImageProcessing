## Konsep dan model

Citra grayscale adalah fungsi diskret $I[r,c]$. Citra RGB menggunakan tiga saluran $I[r,c,k]$. Nilai piksel bukan besaran fisik mutlak tanpa kalibrasi sensor, iluminasi, dan respons kamera.

## Impor pustaka dan cek versi

```python
import numpy as np
import skimage, sklearn
import matplotlib.pyplot as plt
print('NumPy:', np.__version__, 'scikit-image:', skimage.__version__, 'scikit-learn:', sklearn.__version__)
```

## Citra tersedia secara lokal

```python
from skimage import data
img = data.camera()
print('Bentuk:', img.shape, 'tipe:', img.dtype, 'rentang:', (img.min(), img.max()))
plt.figure(figsize=(6, 4)); plt.imshow(img, cmap='gray', vmin=0, vmax=255); plt.axis('off'); plt.show()
```

## Ekstraksi ROI

```python
potongan = img[80:260, 120:320]
fig, ax = plt.subplots(1, 2, figsize=(8, 3))
ax[0].imshow(img, cmap='gray'); ax[0].set_title('Citra penuh')
ax[1].imshow(potongan, cmap='gray'); ax[1].set_title(f'ROI {potongan.shape}')
for a in ax: a.axis('off')
plt.show()
```
